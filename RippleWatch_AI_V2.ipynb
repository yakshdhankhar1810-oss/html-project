```python
# STEP 6: Build reusable TinyFish fetcher

import os
import requests
from datetime import datetime, timezone

TINYFISH_ENDPOINT = "https://api.fetch.tinyfish.ai"


def fetch_article(url):
    """Fetch one article using TinyFish."""

    api_key = os.environ.get("TINYFISH_API_KEY")

    if not api_key:
        raise ValueError(
            "TinyFish API key missing. Run Step 5A first."
        )

    response = requests.post(
        TINYFISH_ENDPOINT,
        headers={
            "X-API-Key": api_key,
            "Content-Type": "application/json"
        },
        json={
            "urls": [url],
            "format": "markdown"
        },
        timeout=120
    )

    response.raise_for_status()
    payload = response.json()

    results = payload.get("results", [])
    errors = payload.get("errors", [])

    if not results:
        return {
            "requested_url": url,
            "title": None,
            "final_url": None,
            "text": "",
            "published_date": None,
            "fetched_at": datetime.now(timezone.utc).isoformat(),
            "status": "fetch_failed",
            "error": str(errors)[:500]
        }

    article = results[0]
    article_text = article.get("text") or ""

    return {
        "requested_url": url,
        "title": article.get("title"),
        "final_url": article.get("final_url"),
        "text": article_text,
        "published_date": article.get("published_date"),
        "fetched_at": datetime.now(timezone.utc).isoformat(),
        "status": "fetched" if article_text.strip() else "empty_content",
        "error": ""
    }


# Test with a harmless example page
test_article = fetch_article("https://example.com")

print("Status:", test_article["status"])
print("Title:", test_article["title"])
print("Final URL:", test_article["final_url"])
print("Text length:", len(test_article["text"]))
print("Published date:", test_article["published_date"])
print("\nSTEP 6 PASSED" if test_article["status"] == "fetched"
      else "STEP 6 NEEDS REVIEW")
```

```python

# STEP 10: Clean the evidence review queue

import pandas as pd
from pathlib import Path

DATA_DIR = Path("/content/ripplewatch_ai_v2/data")

queue = pd.read_csv(DATA_DIR / "evidence_review_queue.csv")

# Remove obvious page clutter and irrelevant text
exclude_phrases = [
    "subscribe to",
    "sign up here",
    "add us on google",
    "copy link",
    "email",
    "linkedin",
    "facebook",
    "filed under:",
    "this audio is auto-generated",
    "please let us know if you have feedback",
    "via getty images",
    "a sign reading"
]

def is_useful_evidence(sentence):
    text = str(sentence).lower()

    if any(phrase in text for phrase in exclude_phrases):
        return False

    # Keep sentences that express an event, impact,
    # observed disruption, or a potential risk.
    evidence_terms = [
        "reported", "disruption", "disruptions",
        "delays", "delayed", "affected", "impact",
        "impacts", "bottlenecks", "constraints",
        "interruptions", "exposed", "expected",
        "anticipated", "throughput", "fulfillment"
    ]

    return any(term in text for term in evidence_terms)

queue["keep_for_review"] = queue["evidence_sentence"].apply(
    is_useful_evidence
)

clean_queue = queue[queue["keep_for_review"]].copy()

clean_queue["review_status"] = "pending_review"
clean_queue["evidence_type"] = "industry_signal"
clean_queue["supplier_specific_alert_allowed"] = False

output_path = DATA_DIR / "evidence_clean_queue.csv"
clean_queue.to_csv(output_path, index=False)

print("Original review queue:", len(queue))
print("Kept for evidence review:", len(clean_queue))
print("Removed as likely page clutter:", len(queue) - len(clean_queue))

print("\n--- CLEAN EVIDENCE QUEUE ---\n")

for _, row in clean_queue.iterrows():
    print(f'{row["candidate_id"]}: {row["evidence_sentence"]}\n')

print("Saved to:", output_path)
print("\nSTEP 10 PASSED")
```

```python

# STEP 11: Classify evidence signals

import pandas as pd
from pathlib import Path

DATA_DIR = Path("/content/ripplewatch_ai_v2/data")

queue = pd.read_csv(DATA_DIR / "evidence_clean_queue.csv")

def classify_evidence(sentence):
    text = str(sentence).lower()

    future_terms = [
        "looking ahead",
        "in the coming years",
        "expected",
        "anticipated",
        "potential disruptions",
        "still anxious",
        "had not experienced"
    ]

    limitation_terms = [
        "did not ask",
        "my sense",
        "likely multiple causes"
    ]

    if any(term in text for term in future_terms):
        return "potential_future_risk"

    if any(term in text for term in limitation_terms):
        return "context_or_uncertainty"

    if any(term in text for term in [
        "reported", "disruptions", "delays",
        "affected", "negative impacts", "exposed"
    ]):
        return "reported_industry_signal"

    return "needs_manual_review"


queue["signal_class"] = queue["evidence_sentence"].apply(
    classify_evidence
)

# Keep the alert permission disabled for all industry-level evidence
queue["supplier_specific_alert_allowed"] = False

output_path = DATA_DIR / "evidence_classified.csv"
queue.to_csv(output_path, index=False)

print("--- CLASSIFICATION COUNTS ---")
print(queue["signal_class"].value_counts().to_string())

print("\n--- CLASSIFIED EVIDENCE ---\n")
for _, row in queue.iterrows():
    print(f'{row["candidate_id"]} | {row["signal_class"]}')
    print(row["evidence_sentence"])
    print()

print("Supplier-specific alerts allowed:",
      queue["supplier_specific_alert_allowed"].sum())
print("Saved to:", output_path)
print("\nSTEP 11 PASSED")
```

```python

# STEP 12: Normalize evidence and flag duplicates

import pandas as pd
import re
from pathlib import Path

DATA_DIR = Path("/content/ripplewatch_ai_v2/data")

df = pd.read_csv(DATA_DIR / "evidence_classified.csv")

def normalize_evidence(sentence):
    text = str(sentence)

    # Remove Markdown heading markers
    text = re.sub(r"#{1,6}\s*", "", text)

    # Remove common article-title prefix when present
    text = re.sub(
        r"^An article from\s+Electronics manufacturers fret over extreme heat disruptions\s*",
        "",
        text,
        flags=re.IGNORECASE
    )

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text

df["clean_evidence_sentence"] = df["evidence_sentence"].apply(
    normalize_evidence
)

# Create normalized text key to identify exact/near-exact repeats
df["dedup_key"] = (
    df["clean_evidence_sentence"]
    .str.lower()
    .str.replace(r"[^a-z0-9]+", " ", regex=True)
    .str.strip()
)

df["duplicate_text"] = df.duplicated(
    subset=["dedup_key"],
    keep="first"
)

print("Total records:", len(df))
print("Duplicate-text records flagged:", int(df["duplicate_text"].sum()))

print("\n--- CLEANED EVIDENCE PREVIEW ---\n")
for _, row in df.iterrows():
    status = "DUPLICATE TEXT" if row["duplicate_text"] else "KEEP"
    print(f'{row["candidate_id"]} | {status}')
    print(row["clean_evidence_sentence"])
    print()

output_path = DATA_DIR / "evidence_normalized.csv"
df.to_csv(output_path, index=False)

print("Saved to:", output_path)
print("\nSTEP 12 PASSED")
```

```python

# STEP 14: Attach original source metadata

import pandas as pd
from pathlib import Path

DATA_DIR = Path("/content/ripplewatch_ai_v2/data")

evidence = pd.read_csv(
    DATA_DIR / "evidence_normalized.csv"
)
articles = pd.read_csv(
    DATA_DIR / "raw_articles.csv"
)

# Build a source lookup using the original requested URL.
# The URL in evidence may be Markdown-formatted, so use
# the original article URL to identify the source.
articles["source_key"] = (
    articles["final_url"]
    .fillna(articles["requested_url"])
    .astype(str)
    .str.strip()
)

def clean_url(value):
    value = str(value).strip()

    # Convert Markdown links like [URL](URL) to plain URLs.
    if value.startswith("[") and "](" in value and value.endswith(")"):
        value = value[1:value.index("](")]

    return value

evidence["source_url_clean"] = evidence["source_url"].apply(clean_url)

# Match evidence to the article by URL, not by title alone.
source_lookup = articles.set_index("source_key").to_dict("index")

def get_source_metadata(url):
    article = source_lookup.get(url, {})
    return pd.Series({
        "verified_source_title": article.get("title"),
        "verified_source_url": article.get("final_url") or article.get("requested_url"),
        "verified_published_date": article.get("published_date"),
        "verified_fetch_status": article.get("status", "source_not_matched")
    })

metadata = evidence["source_url_clean"].apply(get_source_metadata)

evidence = pd.concat([evidence, metadata], axis=1)

# Keep the records review-only until evidence is validated.
evidence["supplier_specific_alert_allowed"] = False

output_path = DATA_DIR / "evidence_with_sources.csv"
evidence.to_csv(output_path, index=False)

print("Evidence records:", len(evidence))
print("Matched source records:",
      int(evidence["verified_source_title"].notna().sum()))
print("Unmatched source records:",
      int(evidence["verified_source_title"].isna().sum()))

print("\n--- SOURCE MATCH PREVIEW ---\n")
print(evidence[[
    "candidate_id",
    "verified_source_title",
    "verified_published_date",
    "verified_fetch_status"
]].to_string(index=False))

print("\nSupplier-specific alerts allowed:",
      int(evidence["supplier_specific_alert_allowed"].sum()))
print("Saved to:", output_path)
print("\nSTEP 14 COMPLETE")
```

```python

# STEP 15: Validate source dates and review eligibility

import pandas as pd
from pathlib import Path
from datetime import datetime, timezone

DATA_DIR = Path("/content/ripplewatch_ai_v2/data")

df = pd.read_csv(DATA_DIR / "evidence_with_sources.csv")

today = datetime.now(timezone.utc).date()

def validate_date(value):
    if pd.isna(value) or not str(value).strip():
        return "MISSING_DATE"

    parsed = pd.to_datetime(value, errors="coerce", utc=True)

    if pd.isna(parsed):
        return "UNPARSEABLE"

    published = parsed.date()

    if published > today:
        return "FUTURE_DATE"

    age_days = (today - published).days

    if age_days <= 30:
        return "RECENT"
    elif age_days <= 90:
        return "OLDER"
    else:
        return "STALE"

df["source_date_status"] = df["verified_published_date"].apply(
    validate_date
)

def assign_review_status(row):
    if row["verified_fetch_status"] != "fetched":
        return "source_fetch_review"

    if row["source_date_status"] in [
        "MISSING_DATE", "UNPARSEABLE", "FUTURE_DATE"
    ]:
        return "date_review_required"

    return "manual_evidence_review"

df["review_status_final"] = df.apply(
    assign_review_status, axis=1
)

# Never enable supplier-specific alerts from industry-only evidence.
df["supplier_specific_alert_allowed"] = False

output_path = DATA_DIR / "evidence_date_review.csv"
df.to_csv(output_path, index=False)

print("--- DATE STATUS COUNTS ---")
print(df["source_date_status"].value_counts().to_string())

print("\n--- REVIEW STATUS COUNTS ---")
print(df["review_status_final"].value_counts().to_string())

print("\n--- RECORD PREVIEW ---")
print(df[[
    "candidate_id",
    "verified_published_date",
    "source_date_status",
    "review_status_final"
]].to_string(index=False))

print("\nSupplier-specific alerts allowed:",
      int(df["supplier_specific_alert_allowed"].sum()))

print("Saved to:", output_path)
print("\nSTEP 15 COMPLETE")
```

```python

import pandas as pd
from pathlib import Path

# STEP 16: INDUSTRY RISK SCORING

base_dir = Path("/content/ripplewatch_ai_v2/data")

input_path = base_dir / "evidence_date_review.csv"
output_path = base_dir / "industry_risk_scores.csv"

# 1. Load input data
if not input_path.exists():
    raise FileNotFoundError(
        f"Input file not found: {input_path}\n"
        "Please check whether the previous step completed successfully."
    )

df = pd.read_csv(input_path)

# 2. Remove duplicate column names, if any
df = df.loc[:, ~df.columns.duplicated()].copy()

# 3. Check required columns
required_columns = ["candidate_id", "signal_class"]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )

# 4. Calculate preliminary industry risk scores
scores = []
reasons_list = []

for _, row in df.iterrows():
    signal_class = str(row["signal_class"]).strip().lower()

    # Use existing reason labels if present; otherwise start empty
    reasons = []

    # Context-only records receive zero
    if signal_class == "context_or_uncertainty":
        score = 0
        reasons.append("context_only")

    else:
        score = 0

        # Apply the existing project signal classifications
        if signal_class == "reported_industry_signal":
            # Determine the reason from available text/labels
            text_parts = []

            for col in [
                "risk_reasons",
                "evidence_sentence",
                "source_title",
                "verified_source_title",
                "signal_type",
                "event_type",
            ]:
                if col in df.columns and pd.notna(row.get(col)):
                    text_parts.append(str(row[col]))

            text = " ".join(text_parts).lower()

            if any(term in text for term in [
                "supply chain", "supply-chain",
                "component shortage", "delivery delay",
                "disruption", "factory shutdown",
                "operational impact"
            ]):
                score += 25
                reasons.append("industry_supply_chain_signal")

            if any(term in text for term in [
                "operational impact", "production stopped",
                "shutdown", "delivery delay", "factory closure"
            ]):
                score += 15
                reasons.append("reported_operational_impact")

            if not reasons:
                reasons.append("general_industry_context")

        elif signal_class == "potential_future_risk":
            text_parts = []

            for col in [
                "risk_reasons",
                "evidence_sentence",
                "source_title",
                "verified_source_title",
                "signal_type",
                "event_type",
            ]:
                if col in df.columns and pd.notna(row.get(col)):
                    text_parts.append(str(row[col]))

            text = " ".join(text_parts).lower()

            if any(term in text for term in [
                "supply chain", "supply-chain",
                "component shortage", "delivery delay",
                "disruption", "factory shutdown"
            ]):
                score += 25
                reasons.append("industry_supply_chain_signal")

            if any(term in text for term in [
                "operational impact", "production stopped",
                "shutdown", "delivery delay", "factory closure"
            ]):
                score += 15
                reasons.append("reported_operational_impact")

            # Cap future-risk scores at 20
            score = min(score, 20)
            reasons.append("future_risk_cap")

        else:
            reasons.append("general_industry_context")

    scores.append(score)
    reasons_list.append(";".join(reasons))

df["industry_risk_score"] = scores
df["risk_reasons"] = reasons_list

# 5. Mark all records as industry-level and preliminary
df["risk_scope"] = "industry_level_only"
df["score_status"] = "preliminary_needs_review"

# 6. Safety gate: no supplier-specific alerts are authorized
df["supplier_specific_alert_allowed"] = False

# 7. Save output
df.to_csv(output_path, index=False)

# 8. Display diagnostics
print("--- INDUSTRY RISK SCORE COUNTS ---")
print(
    df["industry_risk_score"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\n--- SCORE PREVIEW ---")

preview_columns = [
    "candidate_id",
    "signal_class",
    "industry_risk_score",
    "risk_reasons",
    "risk_scope",
    "score_status",
]

print(
    df[
        [col for col in preview_columns if col in df.columns]
    ].to_string(index=False)
)

# Count explicitly allowed values safely
allowed_count = (
    df["supplier_specific_alert_allowed"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("true")
    .fillna(False)
    .sum()
)

print("\nSupplier-specific alerts allowed:", int(allowed_count))
print("Saved to:", output_path)
print("\nSTEP 16 COMPLETE")
```

```python

# STEP 19: Improve industry-level scoring rules

import pandas as pd
from pathlib import Path

DATA_DIR = Path("/content/ripplewatch_ai_v2/data")

df = pd.read_csv(DATA_DIR / "evidence_date_review.csv")

def improved_score(row):
    text = str(row["clean_evidence_sentence"]).lower()
    signal_class = row["signal_class"]

    # Context and uncertainty should not be treated as direct risk.
    if signal_class == "context_or_uncertainty":
        return pd.Series({
            "industry_risk_score_v2": 0,
            "score_reason_v2": "context_or_uncertainty_only"
        })

    # Future risks get a separate, capped preliminary score.
    if signal_class == "potential_future_risk":
        score = 10

        if any(term in text for term in [
            "supplier delivery delays",
            "transportation delays",
            "power reliability",
            "operating constraints",
            "equipment downtime"
        ]):
            score += 10

        return pd.Series({
            "industry_risk_score_v2": min(score, 20),
            "score_reason_v2": "potential_future_risk_not_confirmed"
        })

    # Reported industry signals: score based on explicit evidence.
    score = 0
    reasons = []

    if any(term in text for term in [
        "disruption", "disruptions",
        "interruptions", "delays",
        "affected", "exposed"
    ]):
        score += 10
        reasons.append("reported_disruption_or_exposure")

    if any(term in text for term in [
        "supplier delivery delays",
        "supplier fulfillment",
        "negative impacts",
        "operational throughput",
        "moderate and major"
    ]):
        score += 10
        reasons.append("reported_supply_or_operational_impact")

    if any(term in text for term in [
        "more than half",
        "two-fifths",
        "more than four-fifths",
        "more than two-thirds",
        "50% of manufacturers"
    ]):
        score += 10
        reasons.append("survey_prevalence_statistic")

    return pd.Series({
        "industry_risk_score_v2": min(score, 40),
        "score_reason_v2": ";".join(reasons) if reasons else "low_or_unclear_signal"
    })

scored = df.apply(improved_score, axis=1)
df = pd.concat([df, scored], axis=1)

# Strict safety boundary: industry evidence cannot trigger supplier alerts.
df["supplier_specific_alert_allowed"] = False
df["risk_scope"] = "industry_level_only"
df["score_status"] = "preliminary_needs_review"

output_path = DATA_DIR / "industry_risk_scores_v2.csv"
df.to_csv(output_path, index=False)

print("--- UPDATED SCORE DISTRIBUTION ---")
print(
    df["industry_risk_score_v2"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\n--- ZERO-SCORE RECORDS ---")
print(
    df.loc[df["industry_risk_score_v2"] == 0, [
        "candidate_id", "signal_class", "score_reason_v2"
    ]].to_string(index=False)
)

print("\nSupplier-specific alerts allowed:",
      int(df["supplier_specific_alert_allowed"].sum()))

print("Saved to:", output_path)
print("\nSTEP 19 COMPLETE")
```

```python

# STEP 29: Safely add clean source URLs to dashboard data

import pandas as pd
from pathlib import Path
import re
import shutil

DATA_DIR = Path("/content/ripplewatch_ai_v2/data")
PROJECT_DIR = Path("/content/ripplewatch_ai_v2")

summary_path = DATA_DIR / "dashboard_evidence_summary.csv"
source_path = DATA_DIR / "evidence_with_sources.csv"
app_path = PROJECT_DIR / "app.py"

# Back up the current app before future UI changes
backup_path = PROJECT_DIR / "app_before_source_links.py"
if app_path.exists() and not backup_path.exists():
    shutil.copy2(app_path, backup_path)

summary = pd.read_csv(summary_path)
sources = pd.read_csv(source_path)

def clean_url(value):
    if pd.isna(value):
        return ""
    value = str(value).strip()

    # Convert Markdown links to the URL inside the first [...] segment.
    match = re.search(r"\[(https?://[^\]]+)\]", value)
    if match:
        value = match.group(1)

    # Remove surrounding whitespace and common Markdown punctuation.
    return value.strip().rstrip(")., ")

sources["source_url_cleaned_for_dashboard"] = (
    sources["verified_source_url"].map(clean_url)
)

# Keep one source record per candidate_id before merging.
source_lookup = sources[
    ["candidate_id", "source_url_cleaned_for_dashboard"]
].drop_duplicates(subset=["candidate_id"])

updated = summary.drop(
    columns=["source_url"],
    errors="ignore"
).merge(
    source_lookup,
    on="candidate_id",
    how="left",
    validate="one_to_one"
)

updated = updated.rename(
    columns={"source_url_cleaned_for_dashboard": "source_url"}
)

updated.to_csv(summary_path, index=False)

print("Rows preserved:", len(updated) == len(summary))
print("Source URLs populated:", int(updated["source_url"].notna().sum()))
print("Unique candidate IDs:", updated["candidate_id"].nunique())
print("App backup exists:", backup_path.exists())

print("\nSample cleaned URLs:")
print(updated[["candidate_id", "source_url"]].head(3).to_string(index=False))

print("\nSTEP 29 COMPLETE")
```

```python

# STEP 30: Robustly extract plain URLs and verify them

import pandas as pd
import re
from pathlib import Path

DATA_DIR = Path("/content/ripplewatch_ai_v2/data")
summary_path = DATA_DIR / "dashboard_evidence_summary.csv"

df = pd.read_csv(summary_path)

def extract_plain_url(value):
    if pd.isna(value):
        return ""

    text = str(value).strip()

    # Extract the first actual HTTP(S) URL, including from Markdown links.
    match = re.search(r"https?://[^\s\]\[<>\"']+", text)

    if not match:
        return ""

    url = match.group(0).rstrip(").,;")
    return url

df["source_url"] = df["source_url"].map(extract_plain_url)

# Validate basic URL structure
df["source_url_valid"] = df["source_url"].map(
    lambda url: bool(re.match(r"^https?://[^/\s]+", url))
)

df.to_csv(summary_path, index=False)

print("Total records:", len(df))
print("Plain URLs extracted:", int((df["source_url"] != "").sum()))
print("Structurally valid URLs:", int(df["source_url_valid"].sum()))
print("\nSample URLs:")
print(df[["candidate_id", "source_url", "source_url_valid"]].head(3).to_string(index=False))

print("\nSTEP 30 COMPLETE")
```

```python

# STEP 31: Explicitly remove Markdown link wrappers

import pandas as pd
from pathlib import Path
import re

summary_path = Path(
    "/content/ripplewatch_ai_v2/data/dashboard_evidence_summary.csv"
)

df = pd.read_csv(summary_path)

def clean_markdown_url(value):
    if pd.isna(value):
        return ""

    text = str(value).strip()

    # If this is a Markdown link, extract its destination URL.
    markdown_match = re.search(
        r"\]\((https?://[^)\s]+)\)",
        text
    )
    if markdown_match:
        return markdown_match.group(1).rstrip(".,;")

    # Otherwise extract a plain URL.
    plain_match = re.search(r"https?://\S+", text)
    if plain_match:
        return plain_match.group(0).rstrip("]).,;")

    return ""

df["source_url"] = df["source_url"].map(clean_markdown_url)
df["source_url_valid"] = df["source_url"].map(
    lambda u: u.startswith(("https://", "http://"))
    and "[" not in u
    and "](" not in u
)

df.to_csv(summary_path, index=False)

print("Records:", len(df))
print("Clean URLs:", int(df["source_url_valid"].sum()))
print("\nSample:")
print(df[["candidate_id", "source_url", "source_url_valid"]].head(3).to_string(index=False))
print("\nSTEP 31 COMPLETE")
```

```python

from pathlib import Path
from datetime import datetime, timezone
import pandas as pd
import shutil

data_dir = Path("/content/ripplewatch_ai_v2/data")
source_path = data_dir / "evidence_with_sources.csv"
target_path = data_dir / "evidence.csv"

if not source_path.exists():
    raise FileNotFoundError(f"Missing source file: {source_path}")

src = pd.read_csv(source_path).fillna("")

# Safety check: do not overwrite an already populated target.
if target_path.exists():
    current = pd.read_csv(target_path)
    if len(current) > 0:
        raise RuntimeError(
            f"STOP: {target_path.name} already has {len(current)} rows. "
            "No file was changed."
        )

# Create a timestamped backup of the current target, if it exists.
if target_path.exists():
    stamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    backup_path = data_dir / f"evidence_empty_backup_{stamp}.csv"
    shutil.copy2(target_path, backup_path)
    print("Previous target backed up to:", backup_path)

now = datetime.now(timezone.utc).isoformat()

def clean_url(row):
    value = str(row.get("verified_source_url", "")).strip()
    if not value:
        value = str(row.get("source_url_clean", "")).strip()
    if not value:
        value = str(row.get("source_url", "")).strip()

    # Convert Markdown [label](URL) into the plain URL.
    if value.startswith("[") and "](" in value and value.endswith(")"):
        value = value.split("](", 1)[1][:-1].strip()

    return value

records = []

for _, row in src.iterrows():
    title = str(row.get("verified_source_title", "")).strip()
    if not title:
        title = str(row.get("source_title", "")).strip()

    published = str(row.get("verified_published_date", "")).strip()
    if not published:
        published = str(row.get("source_published_date", "")).strip()

    sentence = str(row.get("clean_evidence_sentence", "")).strip()
    if not sentence:
        sentence = str(row.get("evidence_sentence", "")).strip()

    records.append({
        "evidence_id": str(row.get("candidate_id", "")).strip(),
        "source_title": title,
        "source_url": clean_url(row),
        "source_published_date": published,
        "fetched_at": now,
        "evidence_sentence": sentence,
        "event_type": str(row.get("signal_class", "")).strip(),
        "matched_supplier_names": "",
        "matched_product_names": "",
        "source_date_status": "UNVERIFIED",
        "review_status": "pending_review",
        "risk_score": "",
        "supplier_specific_alert_allowed": False,
    })

rebuilt = pd.DataFrame(records, columns=[
    "evidence_id",
    "source_title",
    "source_url",
    "source_published_date",
    "fetched_at",
    "evidence_sentence",
    "event_type",
    "matched_supplier_names",
    "matched_product_names",
    "source_date_status",
    "review_status",
    "risk_score",
    "supplier_specific_alert_allowed",
])

# Validate before writing.
assert len(rebuilt) == len(src), "Row count mismatch"
assert rebuilt["evidence_id"].is_unique, "Duplicate evidence IDs detected"
assert rebuilt["evidence_sentence"].str.strip().ne("").all(), \
    "One or more evidence sentences are empty"
assert not rebuilt["supplier_specific_alert_allowed"].any(), \
    "Supplier-specific alerts must remain disabled"

rebuilt.to_csv(target_path, index=False)

# Read back the saved file to verify the result.
check = pd.read_csv(target_path)

print("REBUILD COMPLETE")
print("Rows saved:", len(check))
print("Unique evidence IDs:", check["evidence_id"].nunique())
print("Empty evidence sentences:", int(
    check["evidence_sentence"].fillna("").str.strip().eq("").sum()
))
print("Supplier-specific alerts enabled:", int(
    check["supplier_specific_alert_allowed"].astype(str).str.lower().eq("true").sum()
))
print("Review status counts:")
print(check["review_status"].value_counts(dropna=False).to_string())
print("Saved file:", target_path)
print("\nSTEP 49 COMPLETE")
```

```python

from pathlib import Path
import importlib.util
import subprocess
import sys
import time
import socket

project_dir = Path("/content/ripplewatch_ai_v2")
app_path = project_dir / "app.py"

assert app_path.exists(), "app.py not found"

if importlib.util.find_spec("streamlit") is None:
    print("Streamlit is not installed in this runtime.")
    print("Run this installation command in a separate cell:")
    print("%pip install streamlit")
else:
    # Check whether the default Streamlit port is already listening.
    sock = socket.socket()
    port_busy = sock.connect_ex(("127.0.0.1", 8501)) == 0
    sock.close()

    if port_busy:
        print("Port 8501 is already in use.")
        print("An app may already be running; no second process started.")
    else:
        log_path = project_dir / "streamlit_step72.log"
        log_file = open(log_path, "w", encoding="utf-8")

        process = subprocess.Popen(
            [
                sys.executable, "-m", "streamlit", "run", str(app_path),
                "--server.port", "8501",
                "--server.headless", "true",
            ],
            cwd=str(project_dir),
            stdout=log_file,
            stderr=subprocess.STDOUT,
            start_new_session=True,
        )

        time.sleep(3)
        print("Streamlit process ID:", process.pid)
        print("Log file:", log_path)
        print("Process running:", process.poll() is None)

        if process.poll() is not None:
            print("\nStartup log:")
            print(log_path.read_text(encoding="utf-8")[-5000:])

        print("\nIf the process is running, open the app using your Colab port preview.")
        print("STEP 72 COMPLETE — STARTUP ATTEMPT FINISHED")
```
